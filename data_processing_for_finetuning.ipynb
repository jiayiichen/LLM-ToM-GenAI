## Data preprocessing 

```python
import pandas as pd

# Load the dataset
df = pd.read_excel('qwen3_8b_train_results.xlsx')
df.head()
```

```python
excel_file = pd.ExcelFile('qwen3_8b_train_results.xlsx')
sheet_names = excel_file.sheet_names
print(f"Sheet names in the Excel file: {sheet_names}")

# Load the second sheet
df_sheet2 = pd.read_excel('qwen3_8b_train_results.xlsx', sheet_name=sheet_names[1])
print(f"\nSecond sheet preview:")
df_sheet2.head()
```

```python
# Combine the two sheets into one dataframe
df_combined = pd.concat([df, df_sheet2], ignore_index=True)
print(f"Combined dataframe shape: {df_combined.shape}")
df_combined.head()
```

```python
# Check the total number of records
total_records = len(df_combined)
print(f"Total number of records: {total_records}")
```

```python
# Filter rows where qwen3_8b_reasoning starts with '('
starts_with_paren = df_combined['qwen3_8b_reasoning'].str.startswith('(')
count_starts_with_paren = starts_with_paren.sum()
print(f"Number of rows where qwen3_8b_reasoning starts with '(': {count_starts_with_paren}")
```

```python
# Display all rows where qwen3_8b_reasoning starts with '('
rows_starting_with_paren = df_combined[starts_with_paren]
print(f"Rows where qwen3_8b_reasoning starts with '(':\n")
print(rows_starting_with_paren[[ 'qwen3_8b_reasoning']])
```

```python
# Remove lines starting with "(Note:" from qwen3_8b_reasoning column
df_combined['qwen3_8b_reasoning'] = df_combined['qwen3_8b_reasoning'].str.replace(
    r'\(Note:.*?\)', '', regex=True
).str.strip()

print(f"Updated qwen3_8b_reasoning column")
df_combined[['qwen3_8b_reasoning']].head()
```

```python
# check again
# Filter rows where qwen3_8b_reasoning starts with '('
starts_with_paren_new = df_combined['qwen3_8b_reasoning'].str.startswith('(')
count_starts_with_paren_new = starts_with_paren_new.sum()
print(f"Number of rows where qwen3_8b_reasoning starts with '(': {count_starts_with_paren_new}")
```

```python
# Display all rows where qwen3_8b_reasoning starts with '('
rows_starting_with_paren_new = df_combined[starts_with_paren_new]
print(f"Rows where qwen3_8b_reasoning starts with '(':\n")
print(rows_starting_with_paren_new[[ 'qwen3_8b_reasoning']])
```

## Check the total number of failure cases

```python
# Count the number of rows where ANSWER matches qwen3_8b_response
matching_rows = (df_combined['ANSWER'] == df_combined['qwen3_8b_response']).sum()
print(f"Number of rows where ANSWER matches qwen3_8b_response: {matching_rows}")
```

```python
# Filter rows where ANSWER does not match qwen3_8b_response
non_matching_rows = df_combined[df_combined['ANSWER'] != df_combined['qwen3_8b_response']]

# Save to CSV
non_matching_rows.to_csv('qwen3_8b_failure_cases.csv', index=False)

print(f"Number of failure cases (non-matching rows): {len(non_matching_rows)}")
print(f"Saved to 'qwen3_8b_failure_cases.csv'")
```
